```python
area_stats = df.groupby('Area').agg(
    Avg_Price=('Price', 'mean'),
    Total_Restaurants=('Area Restaurant Count', 'max') 
).reset_index()
```

```python
market_avg_price = area_stats['Avg_Price'].mean()
market_avg_comp = area_stats['Total_Restaurants'].mean()
```

```python
print(area_stats.head())
```

```python

plt.figure(figsize=(14, 8))
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=area_stats, 
    x='Total_Restaurants', 
    y='Avg_Price', 
    s=150,               #size of the dots
    color='darkorange',  
    edgecolor='black',   
    alpha=0.8            
)

plt.axhline(y=market_avg_price, color='red', linestyle='--', label='Average Price')
plt.axvline(x=market_avg_comp, color='blue', linestyle='--', label='Average Competition')
plt.legend()
```

```python
gold_mine_areas = area_stats[
    (area_stats['Total_Restaurants'] < market_avg_comp) & 
    (area_stats['Avg_Price'] > market_avg_price)
]

top_opportunities = gold_mine_areas.sort_values(by='Avg_Price', ascending=False)

print("THE BEST CLOUD KITCHEN LOCATIONS \n")
print(top_opportunities.head(5))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ml_df = df.copy()
tier_mapping = {'Budget': 0, 'Mid-Range': 1, 'Premium': 2}
ml_df['Price Tier Encoded'] = df['Price Tier'].map(tier_mapping)
ml_df = ml_df.drop('Price Tier', axis=1)
```

```python
ml_df.head()
```

```python
scaler = StandardScaler()
cols_to_scale = ['Price', 'Menu Count', 'Area Restaurant Count']
ml_df[cols_to_scale] = scaler.fit_transform(ml_df[cols_to_scale])
```

```python
ml_df.head()
```

```python
ml_df.to_csv('cloud_kitchen_ml_ready.csv', index=False)

print("Preprocessing complete! Dataset exported as 'cloud_kitchen_ml_ready.csv'.")
```
